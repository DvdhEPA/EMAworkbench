# EMA Workbench - Run and Save Experiments (Dike Model)

This notebook runs experiments for a specified Dike Network problem formulation using EMA Workbench and saves the results automatically.


```python
# --- Setup and Imports ---
import numpy as np
import scipy as sp
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
import copy

from ema_workbench import (
    ema_logging,
    MultiprocessingEvaluator,
    save_results,
)

from dike_model_function import DikeNetwork
from problem_formulation import (
    get_model_for_problem_formulation,
    sum_over,
    sum_over_time,
)
```

## Setup Environment

Check versions and enable EMA logging.


```python
def setup_environment():
    """Check package versions and configure EMA logging."""
    print(f"Pandas version: {pd.__version__}")
    print(f"NetworkX version: {nx.__version__}")
    ema_logging.log_to_stderr(ema_logging.INFO)


setup_environment()

```

## Load Model for Selected Problem Formulation

Select and load a problem formulation. The formulation number is used later to auto-name the results file.


```python
def load_model(formulation_index):
    """
    Load a specific problem formulation.

    Parameters:
        formulation_index (int): Between 0 and 5

    Returns:
        model: EMA Model
        planning_steps: list
    """
    model, planning_steps = get_model_for_problem_formulation(formulation_index)
    return model, planning_steps


formulation_index = 3
dike_model, planning_steps = load_model(formulation_index)
```

## Inspect Model Parameters

List uncertainties, levers, and outcomes.


```python
def print_model_components(model):
    """Print model uncertainties, levers, and outcomes."""
    print("\nUncertainties:")
    for unc in model.uncertainties:
        print(repr(unc))

    print("\nLevers:")
    for lever in model.levers:
        print(repr(lever))

    print("\nOutcomes:")
    for outcome in model.outcomes:
        print(repr(outcome))


print_model_components(dike_model)
```

## Run Experiments and Save Results

This step runs experiments across a set of scenarios and policies, and saves the results to a file named based on the problem formulation index.


```python
def run_and_save_experiments(model, formulation_index, scenarios, policies):
    """
    Run experiments and save results with a filename indicating the formulation.

    Parameters:
        model: EMA Model
        formulation_index (int): Problem formulation number
        scenarios (int): Number of scenarios
        policies (int): Number of policies
    """
    filename = f"ema_results_problem_{formulation_index}.tar.gz"
    with MultiprocessingEvaluator(model) as evaluator:
        results = evaluator.perform_experiments(
            scenarios=scenarios,
            policies=policies
        )
    save_results(results, filename)
    print(f"Experiments complete. Results saved to: {filename}")


run_and_save_experiments(dike_model, formulation_index,scenarios=1000, policies=300)
```
